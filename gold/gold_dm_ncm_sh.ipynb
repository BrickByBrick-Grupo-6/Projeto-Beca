```python
# Nome do banco de dados onde a tabela será salva
database_name = "dimensao"

# Nome da tabela de destino
table_name = "dm_ncm_sh"

# Caminho completo da tabela no formato database.table
target_path = f"{database_name}.{table_name}"

# Nome da chave primária da tabela
pk = "SK_NCM_SH"
```

```python
# Caminho do diretório Silver para a tabela NCM_SH_CONSOLIDADA no Data Lake
silver_path_sh = f"abfss://silver@stgbbb.dfs.core.windows.net/balancacomercial/NCM_SH_CONSOLIDADA/"
```

```python
# Lê os dados do caminho Delta especificado em 'silver_path_sh' e carrega em um DataFrame Spark chamado 'df_sh'
df_sh = spark.read.format("delta").load(silver_path_sh)
```

```python
# Cria ou substitui uma temporary view chamada "df_ncm_sh" a partir do DataFrame df_sh
df_sh.createOrReplaceTempView("df_ncm_sh")
```

```python
# Consulta para selecionar valores distintos das colunas CO_SH4 e NO_SH4_POR da view temporária df_ncm_sh
query = """
select
 distinct 
    CO_SH4,
    NO_SH4_POR
from df_ncm_sh """
```

```python
# Executa a consulta SQL definida na variável 'query' e armazena o resultado no DataFrame 'df_final'
df_final = spark.sql(query)
```

```python
save_hive_table(df_final, target_path, pk)
```
