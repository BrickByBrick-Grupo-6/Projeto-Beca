```python
# Define o nome do banco de dados onde a tabela será criada ou utilizada
database_name = "dimensao"

# Define o nome da tabela de destino
table_name = "dm_pais"

# Cria o caminho completo da tabela no formato <banco>.<tabela>
target_path = f"{database_name}.{table_name}"

# Define o nome da chave primária da tabela
pk = "PK_PAIS"
```

```python
# Caminho do arquivo Delta contendo os dados consolidados de países na camada Silver do Data Lake
silver_path_paises = f"abfss://silver@stgbbb.dfs.core.windows.net/cnpj/PAISES_CONSOLIDADA/"
```

```python
# Lê a tabela Delta contendo os dados consolidados de países a partir do caminho especificado no Data Lake
df_paises = spark.read.format("delta").load(silver_path_paises)
```

```python
# Cria ou substitui uma view temporária chamada "df_paises" a partir do DataFrame df_paises
df_paises.createOrReplaceTempView("df_paises")
```

```python
# Consulta para selecionar países distintos da tabela temporária df_paises
query = """
select distinct
  codigo_pais,
  descricao_pais
from df_paises
"""
```

```python
# Executa a consulta SQL definida na variável 'query' e armazena o resultado no DataFrame 'df_final'
df_final = spark.sql(query)
```

```python
# Salva o DataFrame df_final na tabela Hive especificada em target_path, utilizando pk como chave primária
save_hive_table(df_final, target_path, pk)
```
