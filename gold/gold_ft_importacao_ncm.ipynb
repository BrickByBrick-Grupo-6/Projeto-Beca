```python
database_name = "fato"
table_name = "ft_importacao_ncm"
target_path = f"{database_name}.{table_name}"
pk = "SK_IMPORTACAO"
```

```python
#Bases utilizadas no relacionamento exportacao por municipio

#Exportacao por municipio
silver_path_s = f"abfss://silver@stgbbb.dfs.core.windows.net/balancacomercial/IMPORTACAO_CONSOLIDADA/"

```

```python
df_imp_ncm = spark.read.format("delta").load(silver_path_s)
```

```python
df_imp_ncm.createOrReplaceTempView("df_imp_ncm")
```

```python
query = """
select
   concat(e.CO_ANO,'-',e.CO_MES) as ano_mes,
    e.CO_NCM,
    e.CO_UNID,

    e.CO_PAIS,
    e.SG_UF_NCM,

    e.QT_ESTAT,
    e.KG_LIQUIDO,
    e.VL_FOB,
    e.VL_FRETE,
    e.VL_SEGURO

from df_imp_ncm e 
"""
```

```python
df_join = spark.sql(query)
```
